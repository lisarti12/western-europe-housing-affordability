```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

pd.set_option(
    "display.max_columns",
    None
)

print("EDA environment ready!")
```

```python
PROJECT_ROOT = Path.cwd().parent

affordability = pd.read_csv(
    PROJECT_ROOT
    / "data"
    / "final"
    / "fact_affordability_annual.csv"
)

market = pd.read_csv(
    PROJECT_ROOT
    / "data"
    / "final"
    / "fact_market_quarterly.csv"
)

print(affordability.head())
print(market.head())
```

```python
latest = (
    affordability
    .sort_values("year")
    .groupby("country_code")
    .tail(1)
)

latest[
    [
        "country",
        "hpi_growth_since_2015",
        "rent_growth_since_2015",
        "income_growth_since_2015",
        "hpi_income_gap_pp",
        "rent_income_gap_pp",
    ]
].sort_values(
    "hpi_income_gap_pp",
    ascending=False
)
```

```python
country = "Netherlands"

subset = affordability[
    affordability["country"] == country
]

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    subset["year"],
    subset["hpi_2015_100"],
    label="House prices"
)

plt.plot(
    subset["year"],
    subset["income_2015_100"],
    label="Disposable income per capita"
)

plt.axhline(
    100,
    linewidth=1
)

plt.title(
    f"House Prices vs Income — {country}"
)

plt.ylabel(
    "Index (2015 = 100)"
)

plt.xlabel("Year")

plt.legend()

plt.show()
```

```python
plt.figure(
    figsize=(10, 6)
)

plt.plot(
    subset["year"],
    subset["rent_2015_100"],
    label="Rent"
)

plt.plot(
    subset["year"],
    subset["income_2015_100"],
    label="Disposable income"
)

plt.axhline(
    100,
    linewidth=1
)

plt.title(
    f"Rent vs Income — {country}"
)

plt.ylabel(
    "Index (2015 = 100)"
)

plt.xlabel("Year")

plt.legend()

plt.show()
```

```python
country = "Germany"

subset_market = market[
    market["country"] == country
].copy()

subset_market["date"] = (
    pd.PeriodIndex(
        subset_market["period"],
        freq="Q"
    )
    .to_timestamp()
)

plt.figure(
    figsize=(11, 6)
)

plt.plot(
    subset_market["date"],
    subset_market["mortgage_rate"]
)

plt.title(
    f"Mortgage Rates — {country}"
)

plt.ylabel(
    "Percent per annum"
)

plt.xlabel("Quarter")

plt.show()
```
